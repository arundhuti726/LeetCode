```python
%sql
CREATE FUNCTION getNthHighestSalary(@N INT) 
RETURNS INT 
AS
--PROCESS 1

-- BEGIN
--     RETURN (
--         /* Write your T-SQL query statement below. */
--     select top 1 salary--max(salary)     max takes 1189 ms, top1 takes 905 ms
--     from(
--             select *, dense_rank() over (order by salary desc) as rank from employee) a
--         where rank=@N
--         )
-- END

--PROCESS 2, 702 ms
BEGIN
RETURN(
        SELECT TOP 1 salary
        FROM
        (
        select *, dense_rank() over (order by salary desc) as rank from employee) a
        where rank=@N

)
END


```
